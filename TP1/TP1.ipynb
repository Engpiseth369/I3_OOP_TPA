Retangle 

```python
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)

    def result(self):
        print("Length of rectangle:", self.length)
        print("Width of rectangle:", self.width)
        print("Area of rectangle:", self.area())
        print("Perimeter of rectangle:", self.perimeter())
        
def main():
    try:
        length = float(input("Enter the length of the rectangle: "))
        width = float(input("Enter the width of the rectangle: "))
        rect = Rectangle(length, width)
        rect.result()
    except ValueError as e:
        print("Error : ", e)

if __name__ == "__main__":
    main()
```

EquationCalculator

```python
class EquationCalculator: #Create class name "EquationCalculator"
    def __init__(self, y, z): #Constructor initializes the values of y and z
        self.y = y #store the input values
        self.z = z 

    def calculate_x(self): #Create the method to 'calculate x" by using the formula
        if self.y == 0 or self.z == 0: #Using condition to identify y, z > 0
             raise ValueError("y and z must not be zero.") #Manually trigger exception(error)
        if self.y + self.z == 0: #Identify y + z > 0
            raise ValueError("y + z must not be zero")
        x = (self.y * self.z)/(self.y + self.z) #Method of equation
        return x #sent the calculate result back to the caller 

    def display_result(self): #Create method to display x
        x = self.calculate_x() #Call function calculate_x
        print(f"x = {x}") #Display value of x

try: #Using try block lets you test a block of code errors
    #Get input from user 
    y = float(input("Enter a value of y: "))
    z = float(input("Enter a value of z: "))

     #Create  calculator object 
    calculator = EquationCalculator(y, z)

    calculator.display_result() #Display the result
except ValueError: #Using except block lets you handle the error
    print("Error: Please enter the valid number!")
```

HundredsCounter

```python
class HundredsCounter:
    """Counts how many complete hundreds are in a non-negative number."""
    def __init__(self, number):
        if number < 0:
            raise ValueError("Number must be non-negative.")

        self.number = number

    def count_hundreds(self):
        return self.number // 100


"""Ask the user for a number and print its complete hundreds."""
try:
    number=int(input("Enter a non-negative Number : "))
    counter = HundredsCounter(number)
    print("Complete Hundreds in the number:", counter.count_hundreds())
except ValueError as e:
    print("Error : ", e)
```

Cheater Game

```python

class LuckyNumberGame:
    def __init__(self, number):
        self.number = number

    def get_next_number(self):
        return self.number + 1

    def get_result(self):
        result = self.get_next_number()
        return result

while True:
    try:
        number = int(input("Please enter an inteter: "))
        game = LuckyNumberGame(number)

    except ValueError:
        print("\n\n")
        print("PLEASE ENTER ONLY A VALID INTEGER!!!!")

    else:
        print("\n\n")
        print(f"The next number is: {game.get_result()}")

    finally:
        print("\n\n")
        print("Do you want to play again? ")
        play_again = ['yes', 'no']
        answer = None
        while answer not in play_again:
            answer = input("Please enter (Yes/No): ").lower()
            if (answer == "no"):
                break
    if answer == 'no':
        break
        
```

# Exercise 6: Integrated Python OOP Practice System
# Reuses the classes from Exercises 1-5. Calculation logic lives in the classes only.

```python
# ---------- Exercise 1: Greeting ----------
class Greeting:
    def __init__(self, name):
        self.name = name
 
    def say_hello(self):
        print("Hello, " + self.name + "!")
 
 
# ---------- Exercise 2: Rectangle ----------
class Rectangle:
    def __init__(self, length, width):
        if length <= 0 or width <= 0:
            raise ValueError("Width and height must be greater than zero.")
        self.length = length
        self.width = width
 
    def area(self):
        return self.length * self.width
 
    def perimeter(self):
        return 2 * (self.length + self.width)
 
    def result(self):
        print("Length of rectangle:", self.length)
        print("Width of rectangle:", self.width)
        print("Area of rectangle:", self.area())
        print("Perimeter of rectangle:", self.perimeter())
 
 
# ---------- Exercise 3: Equation Calculator ----------
class EquationCalculator:
    def __init__(self, y, z):
        self.y = y
        self.z = z
 
    def calculate_x(self):
        if self.y == 0 or self.z == 0:
            raise ValueError("y and z must not be zero.")
        if self.y + self.z == 0:
            raise ValueError("y + z must not be zero.")
        return (self.y * self.z) / (self.y + self.z)
 
    def display_result(self):
        print(f"x = {self.calculate_x()}")
 
 
# ---------- Exercise 4: Hundreds Counter ----------
class HundredsCounter:
    """Counts how many complete hundreds are in a non-negative number."""
 
    def __init__(self, number):
        if number < 0:
            raise ValueError("Number must be non-negative.")
        self.number = number
 
    def count_hundreds(self):
        return self.number // 100
 
 
# ---------- Exercise 5: Cheater Game ----------
class LuckyNumberGame:
    def __init__(self, number):
        self.number = number
 
    def get_next_number(self):
        return self.number + 1
 
    def get_result(self):
        return self.get_next_number()
 
 
# ---------- Menu handlers ----------
def run_greeting():
    name = input("Enter your name: ").strip()
    if not name:
        print("Error: Name cannot be empty.")
        return
    Greeting(name).say_hello()
 
 
def run_rectangle():
    try:
        width = float(input("Enter the width of the rectangle: "))
        height = float(input("Enter the height of the rectangle: "))
        Rectangle(height, width).result()
    except ValueError as e:
        print("Error:", e)
 
 
def run_equation():
    try:
        y = float(input("Enter a value of y: "))
        z = float(input("Enter a value of z: "))
        EquationCalculator(y, z).display_result()
    except ValueError as e:
        print("Error:", e)
 
 
def run_hundreds():
    try:
        number = int(input("Enter a non-negative integer: "))
        counter = HundredsCounter(number)
        print("Complete hundreds in the number:", counter.count_hundreds())
    except ValueError as e:
        print("Error:", e)
 
 
def run_cheater_game():
    try:
        number = int(input("Please enter an integer: "))
        print(f"The next number is: {LuckyNumberGame(number).get_result()}")
    except ValueError:
        print("Error: Please enter only a valid integer!")
 
 
def show_menu():
    print("\n---------- TP02 Python OOP ----------")
    print("1. Greeting")
    print("2. Rectangle")
    print("3. Equation Calculator")
    print("4. Hundreds Counter")
    print("5. Cheater Game")
    print("0. Exit")
    print("-------------------------------------")
 
 
def main():
    actions = {
        "1": run_greeting,
        "2": run_rectangle,
        "3": run_equation,
        "4": run_hundreds,
        "5": run_cheater_game,
    }
 
    while True:
        show_menu()
        choice = input("Enter your choice: ").strip()
 
        if choice == "0":
            print("Goodbye! Thank you for using TP02 Python OOP.")
            break
        elif choice in actions:
            print()
            actions[choice]()
        else:
            print("Invalid choice! Please select a number from 0 to 5.")
 
 
if __name__ == "__main__":
    main()
 
```
